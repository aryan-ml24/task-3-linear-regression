Objective: Implement and understand simple and multiple linear regression using Python, Pandas, Matplotlib and Scikit-learn.

Dataset: `data/tips.csv` (the same dataset used in Task 2, making the workflow consistent). The target variable is `tip`.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load dataset
df = pd.read_csv('data/tips.csv')
print(df.head())
print(df.shape)
```

## 1. Simple Linear Regression

We predict **tip** using one independent variable: **total_bill**.


```python
X = df[['total_bill']]
y = df['tip']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print('Coefficient:', model.coef_[0])
print('Intercept:', model.intercept_)
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f'MAE: {mae:.4f}')
print(f'MSE: {mse:.4f}')
print(f'R²: {r2:.4f}')
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X_test['total_bill'], y_test, alpha=0.7, label='Actual')
order = X_test['total_bill'].argsort().values
plt.plot(X_test['total_bill'].iloc[order], y_pred[order], linewidth=2, label='Regression line')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Simple Linear Regression: Total Bill vs Tip')
plt.legend()
plt.show()
```

### Interpretation
The fitted coefficient is approximately **0.1070**. This means a 1-unit increase in total bill is associated with an estimated increase of about 0.1070 units in tip, on average. The R² score is **0.5449** on the test set.


## 2. Multiple Linear Regression

Now we use two predictors: **total_bill** and **size** to predict the tip amount.


```python
X = df[['total_bill', 'size']]
y = df['tip']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

multi_model = LinearRegression()
multi_model.fit(X_train, y_train)
y_pred = multi_model.predict(X_test)

print('Coefficients:', dict(zip(X.columns, multi_model.coef_)))
print('Intercept:', multi_model.intercept_)
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f'MAE: {mae:.4f}')
print(f'MSE: {mse:.4f}')
print(f'R²: {r2:.4f}')
```

```python
plt.figure(figsize=(7,5))
plt.scatter(y_test, y_pred, alpha=0.7)
mn = min(y_test.min(), y_pred.min())
mx = max(y_test.max(), y_pred.max())
plt.plot([mn, mx], [mn, mx], linewidth=2)
plt.xlabel('Actual Tip')
plt.ylabel('Predicted Tip')
plt.title('Multiple Linear Regression: Actual vs Predicted')
plt.show()
```

## Conclusion

- Simple linear regression models the relationship between one feature and the target.
- Multiple linear regression uses more than one feature.
- MAE and MSE measure prediction error; lower values are better.
- R² measures the proportion of target variance explained by the model; higher values are generally better.
- The coefficients show the direction and estimated size of each feature's effect while the other included features are held constant.
